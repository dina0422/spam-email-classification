### Top Words in Spam vs Ham

```python
nltk.download('stopwords') # Download stopwords
# Initialize stop words
stop_words = set(stopwords.words('english'))

def get_top_words(texts, n=20): # Function to get top n words
    all_words = ' '.join(texts).lower()
    all_words = ''.join([c for c in all_words if c not in string.punctuation])
    words = all_words.split()
    words = [word for word in words if word not in stop_words]
    return Counter(words).most_common(n)

print("Top Spam Words:")
print(get_top_words(df[df['label'] == 1]['text']))

print("\nTop Ham Words:")
print(get_top_words(df[df['label'] == 0]['text']))

```

## Train/Test Split

```python
# 75% train, 25% test
X_train, X_test, y_train, y_test = train_test_split(df.text, df.label , test_size = 0.25) 
```

## CountVectorizer
Builds a vocabulary of all unique tokens in the text corpus

```python
v = CountVectorizer() # Convert text to a matrix of token counts
# Fit the model and transform the training data
X_train_count = v.fit_transform(X_train.values)
X_train_count[:5]
```

## MultinomialNB
Bayes’ Theorem: Naive Bayes classifiers apply Bayes' theorem:![](https://www.google.com/imgres?q=naive%20bayes%20algorithm&imgurl=https%3A%2F%2Fimages.prismic.io%2Fturing%2F65a540077a5e8b1120d5894e_Naive_Bayes_Algorithm_6e504ac475.webp%3Fauto%3Dformat%2Ccompress&imgrefurl=https%3A%2F%2Fwww.turing.com%2Fkb%2Fan-introduction-to-naive-bayes-algorithm-for-beginners&docid=dNtvw-_9RJexcM&tbnid=LZy74C9AomwaUM&vet=12ahUKEwj9qa6Y9IuKAxVrgP0HHb_IOsEQM3oECB0QAA..i&w=1080&h=1080&hcb=2&ved=2ahUKEwj9qa6Y9IuKAxVrgP0HHb_IOsEQM3oECB0QAA)

```python
model = MultinomialNB() # Create a Multinomial Naive Bayes model
# Train the model
model.fit(X_train_count, y_train)
```

```python
X_test_count = v.transform(X_test)
model.score(X_test_count , y_test)

accuracy = model.score(X_test_count , y_test)
print(f"Test Accuracy: {accuracy:.4f}")
```

# Classification report & Confusion Matrix

```python
y_pred = model.predict(X_test_count) # Predict the labels for the test set
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred) # Create confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Ham', 'Spam'], yticklabels=['Ham', 'Spam'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
joblib.dump(model, "spam_classifier_model.pkl") # Save the model
# Save the vectorizer
joblib.dump(v, "vectorizer.pkl")
```
